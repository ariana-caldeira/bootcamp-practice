# 1.4 Sentencias de control

En esta lección aprenderemos que son las sentencias de control y como nos pueden ayudar a gestionar la ejecución de nuestro código.

Las sentencias de control en Python son estructuras que permiten controlar el flujo de ejecución del programa. Estas sentencias, como los condicionales y los bucles, son fundamentales para tomar decisiones y repetir acciones según ciertas condiciones. Son importantes porque nos permiten crear programas más dinámicos y flexibles.

## Sentencia condicional `if-else`

La sentencia condicional `if-else` en Python permite ejecutar bloques de código diferentes dependiendo de una condición dada.

La estructura básica de la sentencia `if-else` es la siguiente:

```python
if condicion:
    # Bloque de código si la condición es verdadera
else:
    # Bloque de código si la condición es falsa
```

La condición es una expresión o comparación que devuelve un valor booleano (`True` o `False`). Si la condición se evalúa como `True`, se ejecutará el bloque de código dentro del `if`. Si la condición se evalúa como `False`, se ejecutará el bloque de código dentro del `else`.

La sentencia condicional `if-else` es importante porque nos permite tomar decisiones en función de condiciones específicas. Esto es especialmente útil en el análisis de datos, donde a menudo necesitamos realizar diferentes acciones según los valores o características de los datos que estamos procesando.

## Sentencia de bucle while

La sentencia de bucle `while` en Python permite repetir un bloque de código mientras se cumpla una condición dada. El bucle continuará ejecutando el código siempre que la condición se evalúe como verdadera.

La estructura básica de un bucle `while` es la siguiente:

```python
while condicion:
    # Bloque de código a repetir mientras la condición sea verdadera
```

Al igual que en las sentencias condicionales `if-else`,la condición es una expresión o comparación que devuelve un valor booleano (`True` o `False`). Si la condición se evalúa como `True`, el bloque de código dentro del bucle `while` se ejecutará. Una vez que el bloque de código se completa, la condición se vuelve a evaluar. Si la condición sigue siendo `True`, el bucle se repetirá y el proceso continuará hasta que la condición se evalúe como `False`. En ese momento, el bucle se detiene y la ejecución del programa continúa con la siguiente instrucción.

Aquí tienes un ejemplo para ilustrar el uso del bucle `while`:

```python
contador = 0

while contador < 5:
    print("El contador es:", contador)
    contador += 1
```

La sentencia de bucle `while` es importante porque nos permite repetir acciones en base a una condición específica, lo que resulta útil cuando no sabemos de antemano cuántas veces se debe repetir el código. Esta estructura de bucle proporciona flexibilidad y control para realizar tareas repetitivas en el análisis de datos.
