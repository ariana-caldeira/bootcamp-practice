# 1.3 Diccionarios y Sets

En esta lección exploraremos otras dos estructuras de datos esenciales, los diccionarios y los sets.

## Diccionarios

Los diccionarios en Python son estructuras de datos que permiten almacenar y organizar información en pares clave-valor. Cada elemento en un diccionario está compuesto por una clave única y su respectivo valor asociado. Los diccionarios son importantes en el análisis de datos debido a su capacidad para acceder y manipular datos eficientemente utilizando claves personalizadas.

Aquí tienes un ejemplo de cómo se ve un diccionario en Python:

```python
# Crear un diccionario de frutas y sus precios
frutas = {"manzana": 0.99, "plátano": 0.75, "naranja": 0.60}

# Acceder a un valor utilizando la clave
precio_manzana = frutas["manzana"]
print("El precio de la manzana es:", precio_manzana)
```

Los diccionarios permiten asociar datos de manera eficiente y personalizada, lo que facilita el acceso y manipulación de información en el análisis de datos. Pueden ser utilizados para organizar metadatos, atributos específicos o realizar mapeos de información, brindando una estructura flexible y poderosa para trabajar con datos en Python.

## Sets

Los sets en Python son estructuras de datos que permiten almacenar una colección de elementos **únicos**, sin orden específico. Los sets se caracterizan por no permitir duplicados, lo que los hace útiles para operaciones como intersecciones, uniones y diferencias de conjuntos. Los sets son importantes en el análisis de datos debido a su capacidad para manejar y manipular conjuntos de datos únicos de manera eficiente.

Aquí tienes un ejemplo de cómo crear y utilizar un set en Python:

```python
# Crear un set de nombres de frutas
frutas = {"manzana", "plátano", "naranja"}

# Agregar un elemento al set
frutas.add("pera")

# Eliminar un elemento del set
frutas.remove("plátano")
```

Los sets en Python son útiles para trabajar con conjuntos únicos de datos y realizar operaciones especializadas. Pueden ser utilizados para eliminar duplicados en una lista, realizar comparaciones entre conjuntos o filtrar elementos duplicados. La capacidad de los sets de manejar conjuntos de datos únicos de manera eficiente los convierte en una herramienta importante en el análisis de datos.
