````python
festivos = {'01/01': 'Año Nuevo', '06/01':'Epifanía del Señor', '15/04':'Viernes Santo', '15/08':'Asunción de la Virgen', '12/10':'Fiesta Nacional de España',
'01/11': 'Todos los Santos', '06/12':'Día de la Constitución Española', '08/12':'Inmaculada Concepción'}
print('1. Este es el diccionario Festivos:', festivos)

monarcas = {'Bélgica': 'Philippe', 'Dinamarca': 'Margrethe II', 'Holanda': 'Beatrix', 'Noruega': 'Harald V', 'España': 'Felipe VI', 'Reino Unido': 'Carlos III'}
print('2. Este es el diccionario Monarcas:', monarcas)

print('3. 15/04 es el festivo de', festivos['15/04'])

monarcas['Holanda']='Willem-Alexander'
print('4. Actualicé el rey de Holanda en el diccionario Monarcas:', monarcas)

rey_luxemburgo = {'Luxemburgo':'Henri'}
monarcas.update(rey_luxemburgo)
print('5. Añadí el rey de Luxemburgo al diccionário:', monarcas)

print('6. Hay', len(monarcas), 'países definidos en el diccionario monarcas')

print('7. ¿Está Monaco entre los paises del diccionario monarcas?', 'Monaco' in monarcas)
print('8. ¿Es verdad que nos olvidamos de poner Suecia?', 'Suecia' in monarcas)

paises = list(monarcas.keys())
print('9. Los países del diccionario monarcas son:', paises)

nombres_festivos = list(festivos.values())
print('10. Los nombres de los festivos son:', nombres_festivos)

print('11. El diccionario festivos es:', festivos)

lista_tuplas = list(festivos.items())
print('12. Lista de tuplas de los elementos del diccionario festivos', lista_tuplas)
````

````python
monarcas2010 = monarcas.copy()
print('13. Copia de monarcas a monarcas2010', monarcas2010)

print('14. Quita España de monarcas2010', monarcas2010.pop('España'))

print('15. Quita el último elemento de monarcas2010', monarcas2010.popitem())

monarcas2010.clear()
print('16. Vacia el diccionario monarcas2010', monarcas2010)
````

````python
print('17. El día 06/01 es festivo de', festivos['06/01'])

print('18. Si no hay festivo 08/03, que en su defecto devuélva', festivos.get('08/03', 'Día Internacional de la Mujer'))

print('19. Si no hay festivo 31/11, que crea un par de clave valor de', festivos.setdefault('31/11', 'Todos los Santos'))
````

## Ejercicios de sets:

````python
temperaturas_diciembre = [10, 10, 13, 14, 12, 15, 
                          12, 11, 13, 17, 18, 15, 
                          15, 17, 17, 13, 13, 11,
                          12, 10, 12, 14, 12, 12,
                          11, 12, 16, 16, 14, 17, 18]

print('1. Define la lista temperaturas_diciembre', temperaturas_diciembre)

diciembre = set(temperaturas_diciembre)
print('2. Conviértala en un set llamado diciembre', diciembre)

colores = {'amarillo', 'negro', 'marrón', 'blanco', 'rosa', 'gris'}
print('3. Crea un set con 6 colores', colores)

colores.add('lila')
print('4. Añade lila', colores)

colores.update({'azul', 'rojo', 'granate'})
print('5. Añade 4 colores más usando .update()', colores)

print('6. El set de colores tiene verde?', 'verde' in colores)
print('7. El set de colores tiene ROJO?', 'ROJO' in colores)

enero = diciembre.copy()
print('8. El set de enero es', enero)

enero.discard(17)
enero.discard(18)
print('9. Quita las temperaturas 17 y 18 de enero', enero)

print('10. En diciembre había', len(diciembre), 'temperaturas distintas')
print('11. En enero había', len(enero), 'temperaturas distintas')

````

````python
print('12. Elementos que hay en diciembre y no están en enero', diciembre.difference(enero))

print('13. Elementos que hay en enero y no están en diciembre', enero.difference(diciembre))

print('14. Elementos que tienen en común enero y diciembre', enero.intersection(diciembre))
````

````python
print('15. Diferencia simétrica entre colores y (marrón,amarillo,cyan) es', colores.symmetric_difference({'marrón','amarillo','cyan'}))
print('La diferencia simétrica excluye los elementos comunes, o sea, devuelve un conjunto de valores que no están en ninguno de los dos sets')
````
