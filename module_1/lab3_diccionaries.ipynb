````python
festivos = {'01/01': 'Año Nuevo', '06/01':'Epifanía del Señor', '15/04':'Viernes Santo', '15/08':'Asunción de la Virgen', '12/10':'Fiesta Nacional de España',
'01/11': 'Todos los Santos', '06/12':'Día de la Constitución Española', '08/12':'Inmaculada Concepción'}
print('1. Este es el diccionario Festivos:', festivos)

monarcas = {'Bélgica': 'Philippe', 'Dinamarca': 'Margrethe II', 'Holanda': 'Beatrix', 'Noruega': 'Harald V', 'España': 'Felipe VI', 'Reino Unido': 'Carlos III'}
print('2. Este es el diccionario Monarcas:', monarcas)

print('3. 15/04 es el festivo de', festivos['15/04'])

monarcas['Holanda']='Willem-Alexander'
print('4. Actualicé el rey de Holanda en el diccionario Monarcas:', monarcas)

rey_luxemburgo = {'Luxemburgo':'Henri'}
monarcas.update(rey_luxemburgo)
print('5. Añadí el rey de Luxemburgo al diccionário:', monarcas)

print('6. Hay', len(monarcas), 'países definidos en el diccionario monarcas')

print('7. ¿Está Monaco entre los paises del diccionario monarcas?', 'Monaco' in monarcas)
print('8. ¿Es verdad que nos olvidamos de poner Suecia?', 'Suecia' in monarcas)

paises = list(monarcas.keys())
print('9. Los países del diccionario monarcas son:', paises)

nombres_festivos = list(festivos.values())
print('10. Los nombres de los festivos son:', nombres_festivos)

print('11. El diccionario festivos es:', festivos)

lista_tuplas = list(festivos.items())
print('12. Lista de tuplas de los elementos del diccionario festivos', lista_tuplas)
````
