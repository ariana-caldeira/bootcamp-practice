# 1.2. Listas y Tuplas

En esta lección exploraremos dos estructuras de datos esenciales, las listas y las tuplas.

## Listas

Las listas son un tipo de estructura de datos que nos permite almacenar una colección de elementos en una sola variable. Cada elemento dentro de una lista está separado por comas y puede ser de cualquier tipo de dato, incluyendo *strings*, *integers*, *floats*, otras listas, etc. Las listas son mutables, lo que significa que podemos modificar sus elementos después de haber sido creadas.

Para crear una lista en Python, se utilizan los corchetes `[]`. Por ejemplo:

```python
temperaturas_diciembre = [10, 10, 13, 14, 12, 15, 12, 11, 13, 17, 18, 15, 15, 17, 17, 13, 13, 11, 12, 10, 12, 14, 12, 12, 11, 12, 16, 16, 14, 17, 18]
```

Podemos acceder a elementos individuales en la lista utilizando estos índices, que comienzan desde cero. Por ejemplo:

| `temperaturas` | 10 | 10 | 13 | ... | 14 | 17 | 18 |
| -------------- | -- | -- | -- | --- | -- | -- | -- |
| índice         | 0  | 1  | 2  | ... | 28 | 29 | 30 |

Si nos fijamos, el índice empieza en `0`, es decir, el primer sitio en la lista tiene el índice `0`. De esta forma el índice `31` **no existe**, aunque la lista tiene 31 contenidos.

Además, podemos agregar elementos nuevos a la lista usando el método `append()`. Por ejemplo:

```python
temperaturas_diciembre.append(23)
```

Existen muchas otras operaciones que se pueden realizar con listas en Python, como eliminar elementos, ordenar elementos, concatenar listas y más. Todos estos métodos los aprenderemos durante esta sesión 💪.

## Tuplas

Las tuplas son una estructura de datos similar a las listas, pero con una diferencia clave: **son inmutables**, lo que significa que no se pueden modificar después de haber sido creadas.

Las tuplas se crean utilizando paréntesis `()` y cada elemento dentro de la tupla está separado por comas. Al igual que con las listas, los elementos dentro de una tupla pueden ser de cualquier tipo de dato. Por ejemplo:

```python
mi_tupla = (1, 'dos', True, 3.14)
```

Para acceder a los elementos de una tupla, se utiliza la misma sintaxis que con las listas, utilizando índices que comienzan en cero. Por ejemplo:

```python
primer_elemento = mi_tupla[0]
segundo_elemento = mi_tupla[1]
```

Sin embargo, dado que las tuplas son inmutables, no podemos modificar los elementos existentes de la tupla o agregar nuevos elementos usando métodos como `append()` que vimos en las listas.

Es importante entender estos conceptos porque nos brindan una forma organizada y eficiente de almacenar y manipular datos en el análisis de datos. Las listas nos permiten realizar operaciones como ordenamiento, filtrado y cálculos estadísticos en conjuntos de datos, mientras que las tuplas nos proporcionan una estructura segura y estable para almacenar información que no debe modificarse.

Al dominar las listas y las tuplas, podrás organizar y manipular datos de manera efectiva, lo que facilitará tareas como la limpieza y preparación de datos, la exploración y visualización de información, y la realización de cálculos y análisis más complejos.
